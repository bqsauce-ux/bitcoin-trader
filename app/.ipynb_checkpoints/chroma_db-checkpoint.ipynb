```python
import chromadb

client = chromadb.PersistentClient(
    path="../data/chroma"
)

collection = client.get_or_create_collection(
    name="project_documents"
)
```

```python
from chroma_db import collection

documents = [
    "Bitcoin uses blockchain technology.",
    "DCA means dollar-cost averaging.",
    "ATR can be used to determine a stop-loss level."
]

ids = [
    "doc_1",
    "doc_2",
    "doc_3"
]

collection.add(
    documents=documents,
    ids=ids
)
```

```python
results = collection.query(
    query_texts=["How does DCA work?"],
    n_results=3
)

print(results)
```
