```python
import pandas as pd


class StrategyMetrics:
    """
    Compare trading activity and performance
    by strategy.
    """

    def __init__(self, results_df):

        self.df = results_df.copy()

    def trade_counts(self):

        trades = self.df[
            self.df["trade"].notna()
        ].copy()

        if trades.empty:
            return {}

        counts = {}

        for trade in trades["trade"]:

            if isinstance(trade, dict):

                strategy = trade.get(
                    "strategy",
                    "UNKNOWN"
                )

                counts[strategy] = (
                    counts.get(strategy, 0) + 1
                )

        return counts

    def strategy_amounts(self):

        trades = self.df[
            self.df["trade"].notna()
        ]

        amounts = {}

        for trade in trades["trade"]:

            if isinstance(trade, dict):

                strategy = trade.get(
                    "strategy",
                    "UNKNOWN"
                )

                amount = float(
                    trade.get("usd_amount", 0)
                )

                amounts[strategy] = (
                    amounts.get(strategy, 0)
                    + amount
                )

        return amounts

    def summary(self):

        return {
            "trade_counts": self.trade_counts(),
            "strategy_amounts": self.strategy_amounts()
        }
```
